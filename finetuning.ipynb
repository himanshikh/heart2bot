```python
from datasets import load_dataset
import os

# ✅ Change these paths to your local directory
# Example: If your project is in "C:\Users\vaish\ESCoT\data\..."
data_dir = r"/home/user/Desktop/PROJECT/ESCoT/data/ablation_data/em_es_ia_sr_re"

dataset = load_dataset(
    "json",
    data_files={
        "train": os.path.join(data_dir, "train", "empathetic_dialogue_train.json"),
        "validation": os.path.join(data_dir, "empathetic_dialogue_valid.json"),
    }
)

# ✅ Access individual splits
train_data = dataset["train"]
val_data = dataset["validation"]

# ✅ Check the first training example
print(train_data[0])

# ✅ Format function for text data
def format_example(example):
    return f"""### Instruction:
{example['instruction']}

### Input:
{example['input']}

### Response:
{example['output']}"""

# ✅ Apply formatting
train_texts = [format_example(x) for x in train_data]
val_texts = [format_example(x) for x in val_data]
```

```python
from transformers import LlamaTokenizer
from datasets import Dataset

# ✅ Load tokenizer
tokenizer = LlamaTokenizer.from_pretrained(
    base_model_id,
    use_fast=False,
    trust_remote_code=True
)

# ✅ Add pad token if missing
if tokenizer.pad_token is None:
    tokenizer.add_special_tokens({'pad_token': tokenizer.eos_token})

from datasets import Dataset

# ✅ Convert formatted text list into Dataset
train_dataset = Dataset.from_dict({"text": train_texts})
val_dataset = Dataset.from_dict({"text": val_texts})

# ✅ Tokenize and align for causal LM training
def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        truncation=True,
        padding="max_length",
        max_length=2048,
    )

tokenized_train = train_dataset.map(tokenize_function, batched=True, remove_columns=["text"])
tokenized_val = val_dataset.map(tokenize_function, batched=True, remove_columns=["text"])

# ✅ Labels must be same as input_ids for language modeling
tokenized_train = tokenized_train.map(lambda samples: {"labels": samples["input_ids"]})
tokenized_val = tokenized_val.map(lambda samples: {"labels": samples["input_ids"]})
```

```python
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training

# ✅ Enable gradient checkpointing to save VRAM
model.gradient_checkpointing_enable()

# ✅ Prepare model for 4-bit (quantized) LoRA training
model = prepare_model_for_kbit_training(model)

# ✅ Define LoRA configuration
lora_config = LoraConfig(
    r=8,
    lora_alpha=64,
    target_modules=[
        "q_proj", "k_proj", "v_proj", "o_proj",
        "gate_proj", "up_proj", "down_proj"
    ],
    bias="none",
    lora_dropout=0.05,
    task_type="CAUSAL_LM"  # important
)

# ✅ Apply LoRA adapters to the model
model = get_peft_model(model, lora_config)

# ✅ OPTIONAL: check trainable parameters
model.print_trainable_parameters()

# ✅ RECOMMENDED (stabilizes memory in local setups)
import torch
torch.cuda.empty_cache()
```

```python
from transformers import DataCollatorForLanguageModeling, TrainingArguments, Trainer
import torch
import os

# ✅ Make sure CUDA is set
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

# ✅ Data collator (for causal LM)
data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

# ✅ Output directories (change paths for your system)
output_dir = "/home/user/Desktop/PROJECT/llama"
log_dir = "/home/user/Desktop/PROJECT/llama"

os.makedirs(output_dir, exist_ok=True)
os.makedirs(log_dir, exist_ok=True)

# ✅ Training arguments
training_args = TrainingArguments(
    output_dir=output_dir,
    per_device_train_batch_size=2,
    gradient_accumulation_steps=2,
    num_train_epochs=1,
    learning_rate=1e-4,
    bf16=False,
    optim="paged_adamw_8bit",
    logging_dir="/content/ESCoT/log",
    save_strategy="epoch",
    logging_steps=10,
    save_total_limit=2,
    report_to="none",
    remove_unused_columns=False,  # ✅ important fix
)


# ✅ Trainer setup
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_val,
    data_collator=data_collator,
)

model.config.use_cache = False
trainer.train()
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

# ✅ Path to your fine-tuned model directory (same as training output_dir)
finetuned_model_path = "/home/user/Desktop/PROJECT/llama/checkpoint-299"

# ✅ Use 8-bit loading to prevent CUDA OOM
bnb_config = BitsAndBytesConfig(
    load_in_8bit=True,
    llm_int8_threshold=6.0,
    llm_int8_has_fp16_weight=False
)

# ✅ Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(finetuned_model_path, use_fast=False)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

# ✅ Load model (on GPU if available)
model = AutoModelForCausalLM.from_pretrained(
    finetuned_model_path,
    quantization_config=bnb_config,
    device_map={"": "cpu"}
)
model.eval()

# -------------------------------
# 🔹 Example test instruction
# -------------------------------
prompt = """### Instruction:
Write an empathetic response to a user who says: 
"I am feeling really lonely these days."

### Input:
None

### Response:
"""

# ✅ Tokenize
inputs = tokenizer(prompt, return_tensors="pt").to("cuda" if torch.cuda.is_available() else "cpu")

# ✅ Generate
with torch.no_grad():
    output = model.generate(
        **inputs,
        max_new_tokens=200,
        temperature=0.7,
        top_p=0.9,
        do_sample=True,
        repetition_penalty=1.2,
        pad_token_id=tokenizer.eos_token_id
    )

# ✅ Decode
response = tokenizer.decode(output[0], skip_special_tokens=True)
print("\n========== Model Response ==========\n")
print(response)
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

# ✅ Path to your fine-tuned model directory
finetuned_model_path = "/home/user/Desktop/PROJECT/llama/checkpoint-299"

# ✅ Use 8-bit loading (safe even if CUDA not used)
bnb_config = BitsAndBytesConfig(
    load_in_8bit=True,
    llm_int8_threshold=6.0,
    llm_int8_has_fp16_weight=False
)

# ✅ Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(finetuned_model_path, use_fast=False)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

# ✅ Load model (CPU-only or GPU if available)
device = "cuda" if torch.cuda.is_available() else "cpu"
model = AutoModelForCausalLM.from_pretrained(
    finetuned_model_path,
    quantization_config=bnb_config,
    device_map={"": device}
)
model.eval()

# -------------------------------
# 🔹 Example CoT Instruction
# -------------------------------
prompt = """### Instruction:
Generate the response as the supporter using the pipeline of Emotion, Emotion Stimulus, Individual Appraisal, Strategy Reason, Response.

### Input:
supporter: Hello, how are you feeling today?
seeker: I'm feeling really lonely these days.

### Response:
"""

# ✅ Tokenize
inputs = tokenizer(prompt, return_tensors="pt").to(device)

# ✅ Generate structured output
with torch.no_grad():
    output = model.generate(
        **inputs,
        max_new_tokens=400,
        temperature=0.7,
        top_p=0.9,
        do_sample=True,
        repetition_penalty=1.2,
        pad_token_id=tokenizer.eos_token_id
    )

# ✅ Decode and print
response = tokenizer.decode(output[0], skip_special_tokens=True)

print("\n========== Model Response ==========\n")
print(response)
```
